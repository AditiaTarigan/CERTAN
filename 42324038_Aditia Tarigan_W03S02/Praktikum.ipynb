```python
import pandas as pd
df = pd.read_csv('Salary_Data.csv')
print(df.head())
```

```python
X = df.iloc[:, :-1].values
y = df.iloc[:, 1].values
```

```python

import matplotlib.pyplot as plt
plt.scatter(X, y, color = "red")
plt.xlabel('Years Experience')
plt.ylabel('Salary')
plt.show()
```

```python
#bagi dataset jadi train dan test 80:20
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

```python
# Fitting Simple Linear Regression to the Training set
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

```python
a = regressor.intercept_
b = regressor.coef_[0]
print("Estimated model intercept, a:", a)
print("Estimated model slope, b:", b)
print(f"ŷ = {a:.2f} + {b:.2f}x")
```

```python
# memprediksi hasil pada data uji (Test set) menggunakan model regresi yang tela
y_pred = regressor.predict(X_test)
y_pred
```

```python
# Menghitung nilai R-Square, yang mengukur seberapa baik model regresi linear
# Semakin mendekati 1, semakin baik model
from sklearn.metrics import r2_score
score=r2_score(y_test,y_pred)
score
```

```python
# Visualizing the Training set results
plt.scatter(X_train, y_train, color = 'red')
plt.plot(X_train, regressor.predict(X_train), color = 'blue')
```

```python
# Visualizing the Test set results
plt.scatter(X_test, y_test, color = 'red')
plt.plot(X_train, regressor.predict(X_train), color = 'blue')
plt.title('Salary vs Experience (Test set)')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

```python
print("Nama:aditia tarigan")
print("NIM:42324038")
print()
import pandas as pd
df = pd.read_csv('rounded_hours_student_scores.csv')
print(df.head())
```

```python
X = df.iloc[:, :-1].values
y = df.iloc[:, 1].values
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

```python
# Memprediksi hasil pada data uji
y_pred = regressor.predict(X_test)

# Mengambil nilai intercept (a) dan slope (b)
a = regressor.intercept_
b = regressor.coef_[0]

print("Estimated model intercept, a:", a)
print("Estimated model slope, b:", b)
print(f"Persamaan: y = {a:.2f} + {b:.2f}x")
```

2.Multiple Linear Regression: prediksi sales (y) berdasarkan data marketing (x) di
youtube, facebook, newspaper, dari dataset
https://www.kaggle.com/datasets/fayejavad/marketing-linear-multiple-regression

```python
import pandas as pd
data_mkt = pd.read_csv('Marketing_Data.csv')
print(data_mkt.head())
```

```python
X_mkt = data_mkt.iloc[:, :-1]
y_mkt = data_mkt.iloc[:, 3]
```

```python
X_train_mkt, X_test_mkt, y_train_mkt, y_test_mkt = train_test_split( X_mkt, y_mkt, test_size=0.2, random_state=0)
```

```python
regressor_mkt = LinearRegression()
regressor_mkt.fit(X_train_mkt, y_train_mkt)
```

```python
y_pred_mkt = regressor_mkt.predict(X_test_mkt)
print("Hasil Prediksi Sales:")
print(y_pred_mkt)
```

```python
a_mkt = regressor_mkt.intercept_
b1_mkt = regressor_mkt.coef_[0]
b2_mkt = regressor_mkt.coef_[1]
b3_mkt = regressor_mkt.coef_[2]

print("Estimated model intercept, a:", a_mkt)
print("Estimated model slope, b1:", b1_mkt)
print("Estimated model slope, b2:", b2_mkt)
print("Estimated model slope, b3:", b3_mkt)
print(f"y = {a_mkt:.2f} + {b1_mkt:.2f}x1 + {b2_mkt:.2f}x2 + {b3_mkt:.2f}x3")
```

```python
score_mkt = r2_score(y_test_mkt, y_pred_mkt)
print("R-Square score:", score_mkt)
```
